### Pandas Analyzer & Data Visualizer


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Libraries Imported Successfully!")
```

```python
data = {
    "SalesID": range(101, 121),
    "Date": pd.date_range("2025-01-01", periods=20),
    "Product": ["Laptop", "Phone", "Tablet", "Headphone", "Keyboard"] * 4,
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Accessories", "Accessories"] * 4,
    "Region": ["North", "South", "East", "West", "Central"] * 4,
    "Sales": [500, 650, 420, 300, 250, 700, 800, 450, 350, 280,
              900, 750, 520, 400, 320, 1000, 850, 600, 450, 380],
    "Profit": [80, 110, 60, 45, 35, 120, 140, 70, 55, 40,
               160, 130, 90, 65, 50, 180, 150, 100, 75, 60],
    "Quantity": [5, 6, 4, 3, 2, 7, 8, 5, 4, 3,
                 9, 7, 6, 4, 3, 10, 8, 6, 5, 4],
    "Year": [2025] * 20
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print("Sales Dataset Created Successfully!")
display(df.head())
```

```python
class SalesDataAnalyzer:

    def __init__(self):
        self.data = None
        self.last_fig = None
        print("Sales Data Analyzer Started!")

    def __del__(self):
        print("Sales Data Analyzer Closed!")

    def load_data(self, file_path):
        try:
            self.data = pd.read_csv(file_path)
            print("Dataset Loaded Successfully!")
            display(self.data.head())

        except FileNotFoundError:
            print("File Not Found!")

        except Exception as e:
            print("Error:", e)

    def check_data(self):
        if self.data is None:
            print("Please Load Dataset First!")
            return False
        return True

    def explore_data(self):
        if not self.check_data():
            return

        while True:
            print("\n1. First 5 Rows")
            print("2. Last 5 Rows")
            print("3. Column Names")
            print("4. Data Types")
            print("5. Basic Information")
            print("6. Describe Data")
            print("7. Back")

            ch = input("Enter Choice: ")

            if ch == "1":
                display(self.data.head())
            elif ch == "2":
                display(self.data.tail())
            elif ch == "3":
                print(self.data.columns.tolist())
            elif ch == "4":
                print(self.data.dtypes)
            elif ch == "5":
                self.data.info()
            elif ch == "6":
                display(self.data.describe(include="all"))
            elif ch == "7":
                break
            else:
                print("Invalid Choice!")

    def clean_data(self):
        if not self.check_data():
            return

        print("Missing Values Before Cleaning:")
        print(self.data.isnull().sum())

        for col in self.data.select_dtypes(include=np.number):
            self.data[col] = self.data[col].fillna(
                self.data[col].mean()
            )

        for col in self.data.select_dtypes(include="object"):
            if self.data[col].isnull().any():
                self.data[col] = self.data[col].fillna(
                    self.data[col].mode()[0]
                )

        self.data = self.data.drop_duplicates()

        print("Data Cleaning Completed!")
        print(self.data.isnull().sum())

    def numpy_operations(self):
        if not self.check_data():
            return

        cols = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()

        print("Numeric Columns:", cols)
        col = input("Enter Column Name: ")

        if col not in cols:
            print("Invalid Column!")
            return

        arr = self.data[col].dropna().to_numpy()

        print("NumPy Array:", arr)
        print("First Element:", arr[0])
        print("Last Element:", arr[-1])
        print("Indexing:", arr[2])
        print("Slicing:", arr[:5])
        print("Shape:", arr.shape)
        print("Dimensions:", arr.ndim)
        print("Sum:", np.sum(arr))
        print("Mean:", np.mean(arr))

    def mathematical_operations(self):
        if not self.check_data():
            return

        cols = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()

        print("Numeric Columns:", cols)
        col = input("Enter Column Name: ")

        if col not in cols:
            print("Invalid Column!")
            return

        arr = self.data[col].to_numpy()

        print("Addition:", arr + 10)
        print("Subtraction:", arr - 10)
        print("Multiplication:", arr * 2)
        print("Division:", arr / 2)

    def combine_data(self):
        if not self.check_data():
            return

        print("\n1. Concatenate")
        print("2. Merge")
        print("3. Join")

        ch = input("Enter Choice: ")

        other = self.data.copy()

        if ch == "1":
            result = pd.concat(
                [self.data, other], ignore_index=True
            )

        elif ch == "2":
            key = input("Enter Common Column: ")

            if key not in self.data.columns:
                print("Column Not Found!")
                return

            result = pd.merge(
                self.data, other,
                on=key,
                suffixes=("_left", "_right")
            )

        elif ch == "3":
            key = self.data.columns[0]

            left = self.data.set_index(key)
            right = other.set_index(key)

            result = left.join(
                right,
                lsuffix="_left",
                rsuffix="_right"
            ).reset_index()

        else:
            print("Invalid Choice!")
            return

        print("Combined DataFrame:")
        display(result.head(10))

    def split_data(self):
        if not self.check_data():
            return

        print("Columns:", self.data.columns.tolist())

        col = input("Enter Column for Splitting: ")

        if col not in self.data.columns:
            print("Invalid Column!")
            return

        groups = self.data.groupby(col)

        for name, group in groups:
            print("\nGroup:", name)
            display(group)

    def search_sort_filter(self):
        if not self.check_data():
            return

        print("\n1. Search")
        print("2. Sort")
        print("3. Filter")

        ch = input("Enter Choice: ")
        col = input("Enter Column Name: ")

        if col not in self.data.columns:
            print("Invalid Column!")
            return

        if ch == "1":
            value = input("Enter Search Value: ")

            result = self.data[
                self.data[col].astype(str).str.contains(
                    value, case=False, na=False
                )
            ]

            display(result)

        elif ch == "2":
            result = self.data.sort_values(
                by=col, ascending=True
            )
            display(result)

        elif ch == "3":
            value = input("Enter Filter Value: ")

            result = self.data[
                self.data[col].astype(str).str.lower()
                == value.lower()
            ]

            display(result)

        else:
            print("Invalid Choice!")

    def aggregate_functions(self):
        if not self.check_data():
            return

        numeric = self.data.select_dtypes(
            include=np.number
        )

        print("SUM:")
        display(numeric.sum())

        print("MEAN:")
        display(numeric.mean())

        print("COUNT:")
        display(numeric.count())

        print("MINIMUM:")
        display(numeric.min())

        print("MAXIMUM:")
        display(numeric.max())

    def statistical_analysis(self):
        if not self.check_data():
            return

        numeric = self.data.select_dtypes(
            include=np.number
        )

        print("Descriptive Statistics:")
        display(numeric.describe())

        print("Standard Deviation:")
        display(numeric.std())

        print("Variance:")
        display(numeric.var())

        print("Percentiles:")
        display(numeric.quantile([0.25, 0.50, 0.75]))

    def create_pivot_table(self):
        if not self.check_data():
            return

        print("Columns:", self.data.columns.tolist())

        index_col = input("Enter Index Column: ")
        value_col = input("Enter Value Column: ")

        if index_col not in self.data.columns:
            print("Invalid Index Column!")
            return

        if value_col not in self.data.columns:
            print("Invalid Value Column!")
            return

        if not pd.api.types.is_numeric_dtype(
            self.data[value_col]
        ):
            print("Value Column Must Be Numeric!")
            return

        pivot = pd.pivot_table(
            self.data,
            index=index_col,
            values=value_col,
            aggfunc=["sum", "mean", "count"]
        )

        display(pivot)

    def advanced_operations(self):
        if not self.check_data():
            return

        print("\n1. Reindex")
        print("2. GroupBy")
        print("3. Transform")

        ch = input("Enter Choice: ")

        if ch == "1":
            result = self.data.reset_index(drop=True)
            result.index = range(1, len(result) + 1)
            display(result)

        elif ch in ["2", "3"]:
            group_col = input("Enter Group Column: ")
            value_col = input("Enter Numeric Column: ")

            if group_col not in self.data.columns:
                print("Invalid Group Column!")
                return

            if value_col not in self.data.columns:
                print("Invalid Value Column!")
                return

            if not pd.api.types.is_numeric_dtype(
                self.data[value_col]
            ):
                print("Select Numeric Column!")
                return

            if ch == "2":
                result = self.data.groupby(
                    group_col
                )[value_col].agg(
                    ["sum", "mean", "count"]
                )

                display(result)

            else:
                result = self.data.copy()

                result["Group_Mean"] = result.groupby(
                    group_col
                )[value_col].transform("mean")

                display(result)

        else:
            print("Invalid Choice!")

    def visualize_data(self):
        if not self.check_data():
            return

        print("\n1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Stack Plot")
        print("7. Seaborn Heatmap")
        print("8. Seaborn Box Plot")
        print("9. Subplots")

        ch = input("Enter Choice: ")

        numeric = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()

        try:
            if ch == "9":
                if len(numeric) < 2:
                    print("Need two numeric columns!")
                    return

                fig, axes = plt.subplots(2, 2, figsize=(10, 7))

                self.data[numeric[0]].plot(
                    kind="bar", ax=axes[0, 0],
                    title="Bar Plot"
                )

                self.data[numeric[1]].plot(
                    kind="line", ax=axes[0, 1],
                    title="Line Plot"
                )

                axes[1, 0].scatter(
                    self.data[numeric[0]],
                    self.data[numeric[1]]
                )
                axes[1, 0].set_title("Scatter Plot")

                axes[1, 1].hist(
                    self.data[numeric[0]].dropna()
                )
                axes[1, 1].set_title("Histogram")

                fig.tight_layout()
                self.last_fig = fig
                plt.show()
                return

            fig, ax = plt.subplots(figsize=(8, 5))

            if ch == "1":
                col = input("Enter Column: ")

                if col not in self.data.columns:
                    print("Invalid Column!")
                    plt.close(fig)
                    return

                if col in numeric:
                    self.data[col].plot(kind="bar", ax=ax)
                else:
                    self.data[col].value_counts().plot(
                        kind="bar", ax=ax
                    )

                ax.set_title("Bar Plot")
                ax.set_xlabel(col)
                ax.set_ylabel("Value")

            elif ch == "2":
                col = input("Enter Column: ")

                if col not in numeric:
                    print("Invalid Column!")
                    plt.close(fig)
                    return

                ax.plot(self.data[col], marker="o", label=col)
                ax.set_title("Line Plot")
                ax.set_xlabel("Records")
                ax.set_ylabel(col)
                ax.legend()

            elif ch == "3":
                x = input("Enter X Column: ")
                y = input("Enter Y Column: ")

                if x not in self.data.columns or y not in self.data.columns:
                    print("Invalid Column!")
                    plt.close(fig)
                    return

                ax.scatter(self.data[x], self.data[y])
                ax.set_title("Scatter Plot")
                ax.set_xlabel(x)
                ax.set_ylabel(y)

            elif ch == "4":
                col = input("Enter Category Column: ")

                if col not in self.data.columns:
                    print("Invalid Column!")
                    plt.close(fig)
                    return

                values = self.data[col].value_counts()

                ax.pie(
                    values,
                    labels=values.index,
                    autopct="%1.1f%%"
                )
                ax.set_title("Pie Chart")

            elif ch == "5":
                col = input("Enter Numeric Column: ")

                if col not in numeric:
                    print("Invalid Column!")
                    plt.close(fig)
                    return

                ax.hist(
                    self.data[col].dropna(),
                    bins=10,
                    edgecolor="black"
                )
                ax.set_title("Histogram")
                ax.set_xlabel(col)
                ax.set_ylabel("Frequency")

            elif ch == "6":
                cols = input(
                    "Enter Numeric Columns separated by comma: "
                ).split(",")

                cols = [c.strip() for c in cols]

                if len(cols) < 2 or not all(c in numeric for c in cols):
                    print("Enter at least two valid numeric columns!")
                    plt.close(fig)
                    return

                ax.stackplot(
                    range(len(self.data)),
                    *[self.data[c].fillna(0) for c in cols],
                    labels=cols
                )

                ax.legend()
                ax.set_title("Stack Plot")
                ax.set_xlabel("Records")

            elif ch == "7":
                corr = self.data.select_dtypes(
                    include=np.number
                ).corr()

                sns.heatmap(
                    corr,
                    annot=True,
                    cmap="coolwarm",
                    ax=ax
                )
                ax.set_title("Seaborn Heatmap")

            elif ch == "8":
                col = input("Enter Numeric Column: ")

                if col not in numeric:
                    print("Invalid Column!")
                    plt.close(fig)
                    return

                sns.boxplot(
                    x=self.data[col],
                    ax=ax
                )
                ax.set_title("Seaborn Box Plot")

            else:
                print("Invalid Choice!")
                plt.close(fig)
                return

            fig.tight_layout()
            self.last_fig = fig
            plt.show()

        except Exception as e:
            print("Visualization Error:", e)

    def save_visualization(self):
        if self.last_fig is None:
            print("Create a Plot First!")
            return

        file_name = input(
            "Enter File Name (example: chart.png): "
        ).strip()

        if not file_name:
            print("File Name Required!")
            return

        if not file_name.lower().endswith(
            (".png", ".jpg", ".jpeg")
        ):
            file_name += ".png"

        self.last_fig.savefig(
            file_name,
            bbox_inches="tight"
        )

        print("Visualization Saved Successfully!")

    def export_results(self):
        if not self.check_data():
            return

        print("\n1. Export Full Data")
        print("2. Export Statistics")

        ch = input("Enter Choice: ")

        if ch == "1":
            self.data.to_csv(
                "cleaned_sales_data.csv",
                index=False
            )
            print("Data Exported Successfully!")

        elif ch == "2":
            self.data.describe(
                include="all"
            ).to_csv("sales_statistics.csv")

            print("Statistics Exported Successfully!")

        else:
            print("Invalid Choice!")

    def show_conclusion(self):
        if not self.check_data():
            return

        print("\n--- PROJECT CONCLUSION ---")
        print("Total Records:", len(self.data))
        print("Total Columns:", len(self.data.columns))

        if "Sales" in self.data.columns:
            print("Total Sales:", self.data["Sales"].sum())
            print("Average Sales:", self.data["Sales"].mean())

        if "Profit" in self.data.columns:
            print("Total Profit:", self.data["Profit"].sum())

        print("Analysis Completed Successfully!")
```
